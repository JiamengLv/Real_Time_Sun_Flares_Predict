# 根据sunpy中的drms下载数据
链接：https://docs.sunpy.org/projects/drms/en/stable/tutorial.html

```python
# hmi.B_720s[2010.05.01_00:00:00_TAI]

import drms
import pandas as pd

client = drms.Client()
import tqdm
```

```python
client.series(r'hmi\.m_')
```

```python
client.pkeys('hmi.m_720s')  
```

```python
client.pkeys('hmi.v_sht_modes')  
```

```python
series_info = client.info('hmi.v_sht_2drls')
series_info.segments
```

```python
data = []

for years in range(2010,2011):
    for month in range(5,6):
        start_date_str = f'{years}.{month:02d}.01_00:00:00_TAI'
        end_date_str =  f'{years}.{(month+1):02d}.01_00:00:00_TAI'
        
        one_keys, one_segments = client.query(f'hmi.B_720s[{start_date_str}-{end_date_str}@12m]', key='T_REC, T_OBS',
                                                    seg='field,azimuth,inclination,disambig')
        print(one_keys,one_segments)
        for i in range(len(one_keys)):
            data_info = {}
            data_info['T_REC'] = one_keys.T_REC[i]
            data_info["field"] = 'http://jsoc.stanford.edu'+ one_segments.field[i]
            data_info["azimuth"] = 'http://jsoc.stanford.edu'+ one_segments.azimuth[i]
            data_info["inclination"] = 'http://jsoc.stanford.edu'+ one_segments.inclination[i]
            data_info["disambig"] = 'http://jsoc.stanford.edu'+ one_segments.disambig[i]

            data.append(data_info)

        
data_df = pd.DataFrame(data)
data_df.drop_duplicates(inplace=True)
data_df.to_csv('./Hmi_B_data_url.csv',index=False)
```
